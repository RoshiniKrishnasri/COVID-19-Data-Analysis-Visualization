```python
import os
import warnings

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

```python
BASE_DIR = os.path.dirname(os.getcwd())

DATA_PATH = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "owid-covid-data (1).csv",
)

print(f"Dataset Path: {DATA_PATH}")
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset Loaded Successfully")
```

```python
print("Dataset Shape:")
print(df.shape)
```

```python
df.head()
```

```python
df.tail()
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.columns = [
    col.strip().lower()
    for col in df.columns
]

print(df.columns.tolist())
```

```python
COLUMN_MAPPING = {
    "location": "country",
    "entity": "country",
    "day": "date",
    "cases": "new_cases",
    "deaths": "new_deaths",
    "vaccinations": "total_vaccinations",
}

for old_col, new_col in COLUMN_MAPPING.items():

    if old_col in df.columns:

        df.rename(
            columns={old_col: new_col},
            inplace=True
        )

print(df.columns.tolist())
```

```python
required_columns = [
    "country",
    "date",
    "new_cases",
    "new_deaths",
    "total_vaccinations",
]

for col in required_columns:

    if col in df.columns:

        print(f"{col} --> PRESENT")

    else:

        print(f"{col} --> MISSING")
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print(df["date"].head())
```

```python
df.isnull().sum()
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows: {duplicates}")
```

```python
df.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully")
```

```python
numeric_columns = [
    "new_cases",
    "new_deaths",
    "total_vaccinations",
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        ).fillna(0)
```

```python
df["country"] = (
    df["country"]
    .astype(str)
    .fillna("Unknown")
)
```

```python
df.describe()
```

```python
unique_countries = df["country"].nunique()

print(f"Total Unique Countries: {unique_countries}")
```

```python
min_date = df["date"].min()

max_date = df["date"].max()

print(f"Start Date: {min_date}")

print(f"End Date: {max_date}")
```

```python
top_cases = (
    df.groupby("country")["new_cases"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_cases
```

```python
top_deaths = (
    df.groupby("country")["new_deaths"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_deaths
```

```python
top_vaccinations = (
    df.groupby("country")["total_vaccinations"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_vaccinations
```

```python
global_trend = (
    df.groupby("date")[
        [
            "new_cases",
            "new_deaths",
            "total_vaccinations"
        ]
    ]
    .sum()
    .reset_index()
)

global_trend.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    global_trend["date"].values,
    global_trend["new_cases"].values
)

plt.title("Global COVID-19 Cases Over Time")

plt.xlabel("Date")

plt.ylabel("New Cases")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    global_trend["date"].values,
    global_trend["new_deaths"].values
)

plt.title("Global COVID-19 Deaths Over Time")

plt.xlabel("Date")

plt.ylabel("New Deaths")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    global_trend["date"].values,
    global_trend["total_vaccinations"].values
)

plt.title("Global COVID-19 Vaccinations Over Time")

plt.xlabel("Date")

plt.ylabel("Vaccinations")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
latest_date = df["date"].max()

latest_df = df[
    df["date"] == latest_date
]

hotspots = latest_df.sort_values(
    by="new_cases",
    ascending=False
).head(10)

hotspots[
    [
        "country",
        "new_cases",
        "new_deaths"
    ]
]
```

```python
PROCESSED_DATA_PATH = os.path.join(
    BASE_DIR,
    "data",
    "processed",
    "cleaned_covid_data.csv"
)

df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(f"Saved To: {PROCESSED_DATA_PATH}")
```
