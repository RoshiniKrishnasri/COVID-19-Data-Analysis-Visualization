```python
import os
import warnings

import pandas as pd
import numpy as np

warnings.filterwarnings("ignore")
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

```python
BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))

RAW_DATA_PATH = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "owid-covid-data (1).csv"
)

PROCESSED_DATA_PATH = os.path.join(
    BASE_DIR,
    "data",
    "processed",
    "cleaned_covid_data.csv"
)

print("Raw Dataset Path:")
print(RAW_DATA_PATH)

print("\nProcessed Dataset Path:")
print(PROCESSED_DATA_PATH)
```

```python
df = pd.read_csv(RAW_DATA_PATH)

print("Dataset Loaded Successfully")
```

```python
print(df.shape)

df.head()
```

```python
print(df.columns.tolist())
```

```python
df.columns = [
    col.strip().lower()
    for col in df.columns
]

print(df.columns.tolist())
```

```python
COLUMN_MAPPING = {
    "location": "country",
    "entity": "country",
    "day": "date",
    "cases": "new_cases",
    "deaths": "new_deaths",
    "vaccinations": "total_vaccinations",
}

for old_col, new_col in COLUMN_MAPPING.items():

    if old_col in df.columns:

        df.rename(
            columns={old_col: new_col},
            inplace=True
        )

print(df.columns.tolist())
```

```python
required_columns = [
    "country",
    "date",
    "new_cases",
    "new_deaths",
    "total_vaccinations",
]

for col in required_columns:

    if col not in df.columns:

        print(f"{col} --> MISSING")

    else:

        print(f"{col} --> PRESENT")
```

```python
for col in required_columns:

    if col not in df.columns:

        if col == "country":

            df[col] = "Unknown"

        elif col == "date":

            df[col] = pd.NaT

        else:

            df[col] = 0

print(df.columns.tolist())
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print(df["date"].head())
```

```python
initial_shape = df.shape

df = df[
    df["date"].notna()
]

final_shape = df.shape

print("Before Removing Invalid Dates:", initial_shape)

print("After Removing Invalid Dates:", final_shape)
```

```python
df.isnull().sum()
```

```python
df["country"] = (
    df["country"]
    .astype(str)
    .replace("nan", "Unknown")
    .fillna("Unknown")
)

df["country"].head()
```

```python
numeric_columns = [
    "new_cases",
    "new_deaths",
    "new_vaccinations",
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    ).fillna(0)

print("Numeric Conversion Completed")
```

```python
df[numeric_columns].dtypes
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows Found: {duplicates}")

df.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully")
```

```python
df.reset_index(
    drop=True,
    inplace=True
)

df.head()
```

```python
for col in numeric_columns:

    negative_count = (df[col] < 0).sum()

    print(f"{col} Negative Values: {negative_count}")
```

```python
for col in numeric_columns:

    df[col] = np.where(
        df[col] < 0,
        0,
        df[col]
    )

print("Negative Values Cleaned")
```

```python
df["country"] = (
    df["country"]
    .str.strip()
    .str.title()
)

df["country"].head()
```

```python
df.sort_values(
    by="date",
    inplace=True
)

df.head()
```

```python
df["total_cases_country"] = (
    df.groupby("country")["new_cases"]
    .cumsum()
)

df[
    [
        "country",
        "date",
        "new_cases",
        "total_cases_country"
    ]
].head()
```

```python
df["total_deaths_country"] = (
    df.groupby("country")["new_deaths"]
    .cumsum()
)

df[
    [
        "country",
        "date",
        "new_deaths",
        "total_deaths_country"
    ]
].head()
```

```python
df["total_vaccinations_country"] = (
    df.groupby("country")["total_vaccinations"]
    .cumsum()
)

df[
    [
        "country",
        "date",
        "total_vaccinations",
        "total_vaccinations_country"
    ]
].head()
```

```python
df["mortality_rate"] = np.where(
    df["total_cases_country"] > 0,
    (
        df["total_deaths_country"]
        / df["total_cases_country"]
    ) * 100,
    0
)

df[
    [
        "country",
        "mortality_rate"
    ]
].head()
```

```python
df["daily_growth_rate"] = (
    df.groupby("country")["new_cases"]
    .pct_change()
    .fillna(0)
) * 100

df[
    [
        "country",
        "daily_growth_rate"
    ]
].head()
```

```python
df["rolling_avg_cases"] = (
    df.groupby("country")["new_cases"]
    .transform(
        lambda x: x.rolling(
            window=7,
            min_periods=1
        ).mean()
    )
)

df[
    [
        "country",
        "rolling_avg_cases"
    ]
].head()
```

```python
latest_date = df["date"].max()

latest_df = df[
    df["date"] == latest_date
]

hotspots = latest_df.sort_values(
    by="new_cases",
    ascending=False
).head(10)

hotspots[
    [
        "country",
        "new_cases",
        "new_deaths"
    ]
]
```

```python
df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print("Cleaned Dataset Saved Successfully")

print(PROCESSED_DATA_PATH)
```
